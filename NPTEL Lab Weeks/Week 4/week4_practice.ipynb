## Mathematical operations on matrices

Ordinary arithmetic operators work element by element.

```r
a <- matrix(1:4, nrow = 2)
b <- matrix(5:8, nrow = 2)
a
b
```

```r
a + b
```

```r
a - b
```

```r
a * b
```

```r
a / b
```

### Matrix multiplication

`*` is elementwise. True matrix multiplication uses `%*%`.

```r
a * b
```

```r
a %*% b
```

### Transpose, determinant, inverse

```r
t(a)
```

```r
det(a)
```

```r
solve(a)
```

```r
a %*% solve(a)
```

### Identity and diagonal

```r
diag(3)
```

```r
diag(a)
```

```r
diag(c(1, 2, 3))
```

### Solving a linear system

```r
A <- matrix(c(2, 1, 1, 3), nrow = 2)
bv <- c(5, 10)
solve(A, bv)
```

## Logical operators

`&` and `|` work element by element. `&&` and `||` compare only the first element and are meant for `if` conditions.

```r
c(TRUE, TRUE, FALSE) & c(TRUE, FALSE, FALSE)
```

```r
c(TRUE, TRUE, FALSE) | c(TRUE, FALSE, FALSE)
```

```r
!c(TRUE, FALSE)
```

```r
TRUE && FALSE
```

```r
TRUE || FALSE
```

### Combining conditions

```r
x > 5 & x < 20
```

```r
x[x > 5 & x < 20]
```

```r
x < 10 | x > 15
```

### xor, any, all

```r
xor(TRUE, FALSE)
```

```r
any(x > 18)
```

```r
all(x > 1)
```

### Logical values behave as 1 and 0

```r
sum(c(TRUE, TRUE, FALSE, TRUE))
```

```r
mean(x > 10)
```
