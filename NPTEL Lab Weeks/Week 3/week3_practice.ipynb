## Calculations with data vectors

Arithmetic on a vector is applied element by element.

```r
x <- c(2, 4, 6, 8, 10)
x
```

```r
x + 5
```

```r
x - 5
```

```r
x / 2
```

```r
y <- c(1, 2, 3, 4, 5)
x + y
```

```r
x - y
```

```r
x * y
```

```r
x / y
```

### Recycling

A shorter vector is reused from the start until it lines up with the longer one.

```r
c(1, 2, 3, 4, 5, 6) + c(10, 20)
```

```r
c(1, 2, 3, 4, 5, 6) * c(1, 2, 3)
```

If the longer length is not a multiple of the shorter one, R still recycles but warns.

```r
c(1, 2, 3, 4, 5) + c(10, 20)
```

## Integer and modulo division

`%/%` returns the whole-number part of a division. `%%` returns the remainder.

```r
17 / 5
```

```r
17 %/% 5
```

```r
17 %% 5
```

```r
c(10, 11, 12, 13, 14) %% 3
```

```r
c(10, 11, 12, 13, 14) %/% 3
```

### Negative numbers

```r
-17 %/% 5
```

```r
-17 %% 5
```

R rounds the integer division downwards, so the remainder takes the sign of the divisor.

### A common use: testing divisibility

```r
x %% 2 == 0
```

## Matrices

A matrix is a vector arranged into rows and columns. R fills it column by column unless told otherwise.

```r
m <- matrix(1:6, nrow = 2)
m
```

```r
matrix(1:6, nrow = 2, byrow = TRUE)
```

```r
matrix(1:6, ncol = 2)
```

### Dimensions

```r
dim(m)
```

```r
nrow(m)
```

```r
ncol(m)
```

### Naming rows and columns

```r
dimnames(m) <- list(c("r1", "r2"), c("c1", "c2", "c3"))
m
```
