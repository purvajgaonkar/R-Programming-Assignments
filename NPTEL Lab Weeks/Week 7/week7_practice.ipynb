## Sequences

A sequence is a set of values following each other in a regular order. The colon operator produces steps of exactly 1.

```r
1:10
```

```r
10:1
```

```r
5:15
```

```r
15:5
```

### Negative values

```r
-1:-10
```

```r
-10:-1
```

```r
-5:-15
```

### Fractional starting points

The step stays at 1, so the fractional part is carried through every value. The sequence stops before it would overshoot the end.

```r
1.23:10
```

```r
1.23:10.54
```

```r
10.54:2.23
```

```r
-1.23:-10
```

```r
-5.23:6
```

## Sequences of dates

`seq()` also works on dates. The `by` argument accepts "day", "week", "month", "quarter" or "year". Dates must be created with `as.Date()` first.

```r
seq(as.Date("2010-01-01"), as.Date("2017-01-01"), by = "years")
```

### Using length instead of an end date

```r
seq(as.Date("2017-01-01"), by = "days", length = 6)
```

```r
seq(as.Date("2017-01-01"), by = "months", length = 6)
```

```r
seq(as.Date("2017-01-01"), by = "years", length = 6)
```

```r
seq(as.Date("2017-01-01"), by = "weeks", length = 6)
```

```r
seq(as.Date("2017-01-01"), by = "quarter", length = 6)
```

### Counting backwards

A negative `by` runs the sequence in reverse, so the start date is the later one.

```r
startdate = as.Date("2016-1-1")
enddate   = as.Date("2017-1-1")
out = seq(enddate, startdate, by = "-1 month")
out
```

## Repeats

`rep()` replicates values. The two main arguments are `times`, which repeats the whole vector, and `each`, which repeats every element in place.

```r
rep(3.5, times = 10)
```

```r
rep(1:4, 2)
```

### times against each

```r
x = 1:4
x
```

```r
rep(x, times = 3)
```

```r
rep(x, each = 3)
```

```r
rep(1:4, each = 2)
```

### Using both together

`each` is always applied first, whichever order the arguments are written in, so both of the following give the same answer.

```r
rep(1:4, each = 2, times = 3)
```

```r
rep(1:4, times = 3, each = 2)
```

### A different count for each element

When `times` is a vector, each element gets its own repeat count. The two lengths must match.

```r
rep(1:4, 2:5)
```

```r
ans = seq(from = 2, to = 8, by = 2)
ans
```

```r
rep(1:4, ans)
```

### Repeating to a fixed output length

`length.out` sets the size of the result. The input is recycled or cut short to fit.

```r
rep(2, length.out = 5)
```

```r
rep(2, length = 5)
```

```r
rep(c(2, 3), length = 5)
```

```r
rep(c(2, 3, 4), length = 5)
```

```r
rep_len(c(2, 3, 4), length.out = 5)
```

### Repeating characters

```r
rep(c("a", "b", "c"), 2)
```

```r
rep(c("apple", "banana", "cake"), 2)
```

```r
rep("apple", length = 5)
```

```r
rep(c("a", "b", "c"), length = 2)
```

```r
rep(c("a", "b", "c"), length = 5)
```

### Repeating a matrix

`rep()` works on the underlying values, so the matrix structure is lost and a plain vector comes back.

```r
x = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = TRUE)
x
```

```r
rep(x, times = 2)
```

## Ordering

This is the one people confuse. `sort()` returns the **values** in order. `order()` returns the **positions** those values came from.

```r
y = c(9, 8, 5, 7, 6)
y
```

```r
order(y)
```

Read it as: the smallest value sits at position 3, the next at position 5, then 4, then 2, then 1.

```r
sort(y)
```

```r
y[order(y)]
```

`y[order(y)]` reproduces `sort(y)` exactly, which is what `order()` is for.

```r
order(y, decreasing = TRUE)
```

### Why order() matters

`order()` lets one variable be arranged by another, which `sort()` cannot do.

```r
name  = c("Ram", "Amy", "Li", "Kate")
marks = c(72, 91, 55, 84)
name[order(marks, decreasing = TRUE)]
```

## Mode

Every object has a mode, describing how it is stored in memory.

```r
mode(2.432)
```

```r
mode(c(3, 4, 5, 6, 7, 8))
```

```r
mode("India")
```

```r
mode(c("India", "CANADA"))
```

### The cases worth remembering

```r
mode(factor(c("UP", "MP")))
```

A factor reports `numeric` because it is stored as integer codes with a separate table of labels, not as the text you see printed.

```r
mode(list("India", "USA"))
```

```r
mode(data.frame(x = 1:2, y = c("India", "USA")))
```

A data frame reports `list`, because that is what it is underneath: a list of equal-length columns.

```r
mode(print)
```

## Lists

A vector, matrix or array can hold only one type of data. A list can hold several, and its elements need not be the same kind of object at all.

```r
list1 = list(1, 2, 3)
list1
```

```r
list2 = list("water", "juice", "lemonade")
list2
```

### Mixed types in one list

```r
mixed = list(25, "India", TRUE, 3.75)
mixed
```

```r
mode(mixed)
```

```r
length(mixed)
```

### Checking the type of each element

```r
for (i in seq_along(mixed)) {
  cat(i, ":", mode(mixed[[i]]), "\n")
}
```

| Sequence, fixed length | `seq(from, to, length.out = n)` |

| Repeat each element | `rep(x, each = n)` |
| Repeat to a length | `rep(x, length.out = n)` |
| Sort values | `sort(x, decreasing = )` |
| Get sorted positions | `order(x)` |
| Storage type | `mode(x)` |
| Create a list | `list(a, b, c)` |
| Sub-list vs element | `L[n]` vs `L[[n]]` |

Three things worth carrying forward:

1. `sort()` gives values, `order()` gives positions. `x[order(x)]` equals `sort(x)`,
   and `order()` is what lets you arrange one variable by another.
2. `mode(factor(...))` is `numeric`, because a factor stores integer codes rather than
   the labels it prints. `mode(data.frame(...))` is `list` for the same kind of reason.
3. With `rep()`, `each` is applied before `times` no matter which order you write them.
