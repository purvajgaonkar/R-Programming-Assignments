## Single bracket against double bracket

This is the one thing about lists that catches people out. `[` returns a **list** containing the selected elements. `[[` returns the **element itself**, taken out of its slot.

```r
list2[1]
```

```r
list2[[1]]
```

### The difference shows up in the type

```r
mode(list2[1])
```

```r
mode(list2[[1]])
```

`list2[1]` is a list of length 1. `list2[[1]]` is a character string. The first cannot be used where a string is expected.

```r
nchar(list2[[1]])
```

### Double bracket takes one element only

`[[` extracts a single slot, so it cannot be given a range.

```r
list2[2:4]
```

## Vector indexing with a condition

A condition inside the brackets keeps only the elements where it is TRUE.

```r
x = 1:10
x
```

```r
x[(x > 5)]
```

```r
x[(x %% 2 == 0)]
```

```r
x[(x %% 2 == 1)]
```

`%%` is the modulo operator, so `x %% 2 == 0` is TRUE for even numbers.

### What the condition actually produces

The brackets are being handed a logical vector, not a set of positions. Seeing it on its own makes the mechanism clear.

```r
x > 5
```

## Vectors with missing observations

```r
x = 1:10
x[5] = NA
x
```

```r
is.na(x)
```

```r
y = x[!is.na(x)]
y
```

`!` negates, so `!is.na(x)` is TRUE wherever the value is present.

### Why this matters

```r
mean(x)
```

```r
mean(y)
```

A single NA makes the whole mean NA. Removing it first, or using `na.rm = TRUE`, avoids that.

```r
mean(x, na.rm = TRUE)
```

## Creating factors

`factor()` encodes a vector of discrete values. Each distinct value becomes a level, and levels are numbered from 1.

```r
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
x
```

```r
levels(x)
```

The levels come out alphabetically by default: juice, lemonade, water.

### Converting an existing vector

```r
x = c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)
x
```

```r
y = as.factor(x)
y
```

```r
levels(y)
```

Numeric values sort numerically, so the levels run 1 to 6.

---
## Factors with explicit levels and labels

When the observed data contains only some of the possible categories, the full set has to be supplied. Rolling a die seven times may never produce a six, but six is still a valid outcome.

```r
y = c(1, 4, 3, 5, 4, 2, 4)
y
```

```r
possible.dieface = c(1, 2, 3, 4, 5, 6)
labels.dieface   = c("one", "two", "three", "four", "five", "six")
```

```r
facy = factor(y, levels = possible.dieface, labels = labels.dieface)
facy
```

Levels one to six all appear even though six was never rolled. Without the `levels` argument that category would vanish, and a table of results would silently omit it.

```r
table(facy)
```

```r
table(factor(y))
```

The second table has no entry for six at all. That is the practical reason for declaring levels.

## class()

Every object has a class. For simple vectors it matches the mode, but for factors and data frames it does not.

```r
class(9)
```

```r
class("9")
```

```r
class(print)
```

```r
x = matrix(nrow = 2, ncol = 2, data = 1:4)
class(x)
```

### Where class and mode disagree

```r
f = factor(c("A", "B", "A"))
class(f)
```

```r
mode(f)
```

`class` reports `factor`, `mode` reports `numeric`. The class governs how it prints and behaves; the mode reveals that it is stored as numbers underneath.

---
## unclass()

`unclass()` strips the class attribute and exposes what is actually stored.

```r
brands = c("A", "A", "B", "B", "B", "B", "C")
brands
```

```r
brands_fac = factor(brands)
brands_fac
```

```r
unclass(brands_fac)
```

The factor is an integer vector, 1 1 2 2 2 2 3, with a `levels` attribute holding the labels. That is the whole of what a factor is.

### Using the codes as an index

```r
colours = c("blue", "green", "red")
colours[unclass(brands_fac)]
```

Because the codes are 1, 2 and 3, they index straight into another vector. A maps to blue, B to green, C to red.

### Another example

```r
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
unclass(x)
```

Codes are 2 2 1 2 3, because the alphabetical levels put juice first.

---
## Controlling the level order

The `levels` argument sets the order explicitly, which changes the codes assigned.

```r
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"),
           levels = c("water", "juice", "lemonade"))
x
```

```r
unclass(x)
```

```r
levels(x)
```

Same data, different codes: 3 3 2 3 1 instead of 2 2 1 2 3. The order of levels is a choice, not a property of the data.

---
## Ordered factors

`ordered()` records that the categories have a rank, which `factor()` does not.

```r
income = ordered(c("high", "high", "low", "medium", "medium"),
                 levels = c("low", "medium", "high"))
income
```

```r
unclass(income)
```

The levels print with `<` between them, marking the ordering.

### What ordering allows

```r
income > "low"
```

```r
plain = factor(c("high", "low", "medium"), levels = c("low", "medium", "high"))
plain > "low"
```

Comparison works on an ordered factor and fails with a warning on a plain one.

---
## The classic factor trap

This is the single most common factor bug in R, and it follows directly from the storage shown by `unclass()`.

```r
f = factor(c(10, 20, 30, 20))
f
```

```r
as.numeric(f)
```

The answer is 1 2 3 2, not 10 20 30 20. `as.numeric()` returns the internal codes, because that is what is stored.

### The correct conversion

```r
as.numeric(as.character(f))
```

```r
as.numeric(levels(f))[f]
```

Both recover the real values. Go through the labels, never straight to numeric.

## Use of width

`width` pads every string to the same length, which is what lines values up in a column.

```r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "centre")
```

```r
format(c("A", "BB", "CCC", "DDDD"), width = 14, justify = "centre")
```

```r
format("apple")
```

```r
cat("apple", "banana", "\n")
```

```r
cat("Total:", format(1234567, big.mark = ","), "\n")
```

### Building a formatted table

```r
items  = c("Water", "Juice", "Lemonade")
prices = c(15, 120, 4500)

for (i in seq_along(items)) {
  cat(format(items[i], width = 12, justify = "left"),
      format(prices[i], big.mark = ",", width = 8),
      "\n")
}
```

## The five things most likely to be tested

1. `[` returns a list, `[[` returns the element. `mode(L[1])` is `"list"`,
   `mode(L[[1]])` is the element's own type.
2. `unlist()` on a mixed list coerces everything to character, because a vector holds
   one type only.
3. A factor is an integer vector plus a `levels` attribute, which is exactly what
   `unclass()` reveals. Every other factor behaviour follows from this.
4. `as.numeric()` on a factor returns the level codes, not the values. Use
   `as.numeric(as.character(f))`.
5. `format()` returns a character string rather than printing, which is why it nests
   inside `cat()` and `print()`.
