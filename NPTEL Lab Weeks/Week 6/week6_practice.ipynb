### seq_along is safer than 1:length

On an empty vector `length(x)` is 0, and `1:0` counts backwards as `1 0`, so the loop runs twice instead of never.

```r
empty <- c()
1:length(empty)
```

```r
seq_along(empty)
```

```r
for (i in 1:length(empty)) {
  cat("should never print, i =", i, "\n")
}
```

```r
for (i in seq_along(empty)) {
  cat("should never print, i =", i, "\n")
}
```

### Collecting results

Growing a vector with `c()` copies the whole thing on every iteration. Creating it at full size first is much faster.

```r
squares <- c()
for (i in 1:5) {
  squares <- c(squares, i^2)
}
squares
```

```r
squares <- numeric(5)
for (i in 1:5) {
  squares[i] <- i^2
}
squares
```

```r
total <- 0
for (i in 1:10) {
  total <- total + i
}
total
```

## while loop

A `for` loop runs a known number of times. A `while` loop runs as long as its condition stays TRUE.

```r
i <- 1
while (i <= 5) {
  print(i)
  i <- i + 1
}
```

The counter update is essential. Without it the condition never turns FALSE and the loop never ends.

```r
x <- 100
while (x >= 1) {
  cat(x, "")
  x <- x / 2
}
```

## repeat loop

`repeat` has no condition, so it runs forever unless you `break` out of it.

```r
i <- 1
repeat {
  print(i)
  i <- i + 1
  if (i > 5) break
}
```

## break and next

```r
for (i in 1:10) {
  if (i > 3) break
  print(i)
}
```

```r
for (i in 1:10) {
  if (i %% 2 == 0) next
  print(i)
}
```

```r
values <- c(4, 7, 2, -5, 9, 1)

for (i in seq_along(values)) {
  if (values[i] < 0) {
    cat("First negative at position", i, "\n")
    break
  }
}
```

### Returning several values in a list

```r
summary_stats <- function(x) {
  list(n = length(x),
       mean = mean(x),
       median = median(x),
       sd = sd(x))
}

result <- summary_stats(c(2, 4, 4, 4, 5, 5, 7, 9))
result
```

```r
result$mean
```

### Scope

A variable created inside a function is local to it.

```r
x <- 100

test_scope <- function() {
  x <- 5
  cat("inside, x =", x, "\n")
}

test_scope()
cat("outside, x =", x, "\n")
```

### Functions calling functions

```r
is_even <- function(n) {
  n %% 2 == 0
}

count_evens <- function(v) {
  count <- 0
  for (value in v) {
    if (is_even(value)) count <- count + 1
  }
  count
}

count_evens(c(1, 2, 3, 4, 5, 6, 7, 8))
```

```r
factorial_loop <- function(n) {
  result <- 1
  for (i in 1:n) {
    result <- result * i
  }
  result
}

factorial_loop(5)
```

```r
factorial(5)
```

### Recursion

```r
factorial_recursive <- function(n) {
  if (n <= 1) return(1)
  n * factorial_recursive(n - 1)
}

factorial_recursive(5)
```
