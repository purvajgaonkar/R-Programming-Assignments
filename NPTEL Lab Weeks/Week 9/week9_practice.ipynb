## The limitation of print()

The goal is to print one line: *The zero occurs at 2\*pi radians.*

`print()` handles only one object at a time. Passing three raises an error, because the second and third arguments are matched to `print()`'s own parameters rather than printed. The error is caught here so the notebook keeps running.

```r
tryCatch(
  print("The zero occurs at", 2*pi, "radians."),
  error = function(e) cat("Error:", conditionMessage(e), "\n")
)
```

### One at a time

```r
print("The zero occurs at"); print(2*pi); print("radians")
```

It works, but gives three separate lines with quotes and index markers.

## cat()

`cat()` converts its arguments to character, joins them in sequence, and writes a single continuous output. It puts a space between items by default and does not add a newline, so `\n` has to be supplied.

```r
cat("The zero occurs at", 2*pi, "radians.", "\n")
```

### Without the newline

```r
cat("no newline here")
cat("so this runs on")
```

```r
cat("with a newline\n")
cat("this starts fresh\n")
```

### Mixing text and calculated values

```r
d = date()
cat("Today's date is:", d, "\n")
```

```r
x = 7
cat("The square of", x, "is", x^2, "!\n")
```

```r
cat("The square root of", x, "is approximately", format(sqrt(x), digits = 3), "\n")
```

`format()` returns a string, so it nests inside `cat()` to control the digits.

### Printing a vector

```r
evenno = c(2, 4, 6, 8, 10)
evenno
```

```r
cat("The first few even numbers are:", evenno, "...\n")
```

## paste()

`paste()` joins strings end to end and **returns** the result, so unlike `cat()` it can be assigned to a variable. It converts arguments with `as.character()` and inserts a space by default.

```r
paste("Everybody", "loves", "R Programming.")
```

### Changing the separator

```r
paste("Everybody", "loves", "R Programming.", sep = "*")
```

```r
paste("Everybody", "loves", "R Programming.", sep = "===")
```

```r
paste("Everybody", "loves", "R Programming.", sep = "")
```

### paste converts to character

```r
paste(1:12)
```

```r
as.character(1:12)
```

On a single vector the two are identical.

### Storing the result

```r
msg = paste("The answer is", 42)
msg
```

```r
class(msg)
```

## paste() with vectors

When an argument is a vector, `paste()` works term by term and returns a vector. Shorter arguments are recycled.

```r
names = c("Prof. Singh", "Mr. Venkat", "Dr. Jha")
names
```

```r
paste(names, "is", "a good", "person.")
```

```r
length(paste(names, "is", "a good", "person."))
```

---
## collapse

`collapse` takes the vector that `paste()` produced and joins its elements into one string, using the given separator.

```r
paste(names, "is", "a good", "person.", collapse = ", and ")
```

```r
length(paste(names, "is", "a good", "person.", collapse = ", and "))
```

### sep against collapse

`sep` goes between the arguments within each element. `collapse` goes between the finished elements.

```r
x = paste("Ex", 1:5, sep = "_")
x
```

```r
x[1]
```

```r
x[3]
```

```r
x[5]
```

```r
x = paste("Ex", 1:5, sep = "_", collapse = "")
x
```

```r
x[1]
```

```r
length(x)
```

With `collapse` the result is a single string of length 1, so `x[1]` is the whole thing.

```r
paste("Ex", 1:5, sep = "_", collapse = " | ")
```

```r
paste(1, " is first ", 2, " is second ", 3, " is third ", sep = "#")
```

## paste0()

`paste0()` is `paste()` with `sep = ""`, so nothing is inserted between arguments.

```r
paste0(1:10)
```

```r
paste(1:10)
```

On a single vector the two agree.

### Where they differ

```r
paste0(1:10, c("st", "nd", "rd", rep("th", 7)))
```

```r
paste(1:10, c("st", "nd", "rd", rep("th", 7)))
```

`paste()` puts a space between the number and its suffix. `paste0()` does not.

```r
paste0("file_", 1:3, ".csv")
```

## strsplit()

`strsplit(x, split)` breaks each element of a character vector into pieces at the split string. It always returns a **list**, one element per input string.

```r
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"
x
```

### Splitting on one character

```r
strsplit(x, split = "!")
```

### Splitting on a string

```r
strsplit(x, split = "&!")
```

The split string is matched as a whole. `&!` and `!&` are different separators, and the result changes accordingly.

```r
y = strsplit(x, split = "!&")
y
```

### Accessing the pieces

The result is a list, so `[[1]]` reaches the vector inside it, and `[n]` then picks a piece.

```r
y[[1]][1]
```

```r
y[[1]][2]
```

```r
y[[1]][3]
```

Only two pieces exist, so asking for a third returns NA rather than an error.

```r
class(y)
```

```r
length(y[[1]])
```

### Splitting into single characters

```r
strsplit("R course", split = "")
```

---
## strsplit() on a vector, then into a matrix

```r
dates = c("2020-07-24", "2021-08-25", "2022-09-26", "2023-10-27")
dates
```

```r
datesplt = strsplit(dates, "-")
datesplt
```

Four input strings give a list of four, each holding year, month and day.

### Flattening into a character matrix

```r
datemat = matrix(unlist(datesplt), nrow = 4, ncol = 3, byrow = TRUE)
datemat
```

`byrow = TRUE` is essential. `unlist()` produces year, month, day, year, month, day and so on, so filling by row keeps each date on its own row.

```r
class(datemat[1, 1])
```

### As a numeric matrix

```r
datematrix = matrix(as.numeric(unlist(datesplt)), nrow = 4, ncol = 3, byrow = TRUE)
datematrix
```

```r
class(datematrix[1, 1])
```

`as.numeric()` turns `"07"` into `7`, dropping the leading zero, and makes arithmetic possible.

```r
colnames(datematrix) = c("Year", "Month", "Day")
datematrix
```

```r
datematrix[, "Year"] + 1
```

## nchar(): counting characters

`nchar()` returns the number of characters in each element. Spaces and punctuation count.

```r
x = "R course 24.07.2022"
y = "Number of participants: 25"
```

```r
nchar(x)
```

```r
nchar(y)
```

### On a vector

```r
nchar(c("Apple", "Banana", "Cake"))
```

### On numbers

Numbers are converted to character first, so `nchar()` counts digits and the decimal point.

```r
nchar(c(2, 4, 6))
```

```r
nchar(c(11, 222, 3333))
```

```r
nchar(c(1.1, 2.22, 3.333))
```

`1.1` has three characters: 1, the point, and 1.

### nchar is not length

```r
fruits = c("Apple", "Banana", "Cake")
length(fruits)
```

```r
nchar(fruits)
```

`length()` counts elements. `nchar()` counts characters within each element.

---
## nzchar(): non-empty strings

`nzchar()` returns TRUE for each element that is not an empty string. It is a fast check for blanks.

```r
nzchar(x)
```

```r
nzchar(y)
```

```r
nzchar(c("Apple", "Banana", "Cake"))
```

```r
z = c("Apple", "", "Cake")
nzchar(z)
```

```r
z[nzchar(z)]
```

Used as an index it drops the empty strings.

### A space is not empty

```r
nzchar(c("", " "))
```

## toupper() and tolower()

Convert case. Characters that are not letters are left as they are.

```r
x = "R course will start from 24.07.2022"
toupper(x)
```

```r
z = "INDIAN INSTITUTE OF TECHNOLOGY"
tolower(z)
```

```r
toupper(c("apple", "banana", "cake"))
```

### Case-insensitive comparison

```r
"Apple" == "apple"
```

```r
tolower("Apple") == tolower("apple")
```

---
## Putting it together

```r
students = c("amy", "ram", "li")
marks = c(91, 72, 88)

for (i in seq_along(students)) {
  cat(format(toupper(students[i]), width = 6),
      "scored", marks[i],
      "(", nchar(students[i]), "letters in name )\n")
}
```

```r
report = paste0(toupper(students), ":", marks)
report
```

```r
paste(report, collapse = "; ")
```

## Summary

| Task | Command |
|---|---|
| Print several items on one line | `cat(a, b, c, "\n")` |
| Change the separator | `cat(x, sep = " / ")` |
| Label wrapped lines | `cat(x, fill = n, labels = )` |
| Join strings and keep the result | `paste(a, b, sep = " ")` |
| Join with no separator | `paste0(a, b)` |
| Join a vector into one string | `paste(x, collapse = ", ")` |
| Split a string | `strsplit(x, split = "-")` |
| Reach a piece | `strsplit(x, "-")[[1]][2]` |
| Split results into a matrix | `matrix(unlist(s), nrow = , byrow = TRUE)` |
| Count characters | `nchar(x)` |
| Test for non-empty | `nzchar(x)` |
| Change case | `toupper(x)`, `tolower(x)` |

Four things worth carrying forward:

1. `cat()` writes output and returns NULL. `paste()` returns a string you can store.
2. `sep` goes between arguments inside each element; `collapse` joins the finished
   elements into one string of length 1.
3. `strsplit()` always returns a list, so reach the pieces with `[[1]][n]`.
4. `nchar()` counts characters within each element; `length()` counts elements.
