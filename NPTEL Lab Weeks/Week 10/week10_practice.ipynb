## sub(): replace the first match

`sub(old, new, string)` replaces the **first** occurrence of `old` only.

```r
y = "Number of participants: 25"
sub("25", "30", y)
```

```r
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."
y
```

```r
sub("Mr. Singh", "Professor Jha", y)
```

Only the first "Mr. Singh" changed. The second is untouched.

---
## gsub(): replace every match

`gsub()` is a global replace: every occurrence is changed.

```r
gsub("Mr. Singh", "Professor Jha", y)
```

### Side by side

```r
s = "Average temperature of August is 35 ofcourse."
```

```r
sub("of", "in", s)
```

```r
gsub("of", "in", s)
```

`gsub()` also changes the "of" hidden inside "ofcourse". Matching is on characters, not whole words.

### Replacing with nothing

```r
gsub(" ", "", "R course 24.07.2022")
```

## grep(): which elements match

`grep(pattern, x)` searches each element of `x` and returns the **indices** of those that match.

```r
str = c("R Course", "exercises", "include examples of R language")
```

```r
grep("ex", str)
```

```r
grep("ex", str, value = FALSE)
```

### value = TRUE returns the elements instead

```r
grep("ex", str, value = TRUE)
```

### Case sensitivity

```r
str = c("R Course", "exercises", "include examples of r language", "in R software.")
```

```r
grep("R", str, ignore.case = FALSE, value = TRUE)
```

```r
grep("R", str, ignore.case = TRUE, value = TRUE)
```

"exercises" matches once case is ignored, because it contains a lowercase r.

```r
grep("R", str, ignore.case = TRUE, value = FALSE)
```

```r
grep("R", str, ignore.case = FALSE, value = FALSE)
```

### Searching across two strings

```r
x = "R course 24.07.2021"
y = "Number of participants: 25"
c(x, y)
```

```r
grep("our", c(x, y))
```

```r
grep("Num", c(x, y))
```

### Character ranges

Square brackets match any one character in the range.

```r
grep("[f-j]", letters)
```

```r
grep("[f-j]", letters, value = TRUE)
```

### A near-miss

```r
grep("tion", c("caution", "safety", "illusion", "security", "vacation"))
```

"illusion" ends in "sion", so it does not match "tion".

## grepl(): TRUE or FALSE for each element

`grepl()` returns a logical vector, one value per element, rather than positions.

```r
str = c("R Course", "exercises", "include examples of R language")
grepl("R", str)
```

```r
grepl("ex", str)
```

### grepl() has no value argument

The slides show `grepl("ex", str, value = T)`. That argument belongs to `grep()` only, and current R rejects it. The error is caught so the notebook keeps running.

```r
tryCatch(
  grepl("ex", str, value = TRUE),
  error = function(e) cat("Error:", conditionMessage(e), "\n")
)
```

### grepl() as a filter

```r
str[grepl("ex", str)]
```

```r
sum(grepl("ex", str))
```

## attach() and detach()

`attach()` makes the columns available by name, without the `painters$` prefix.

```r
attach(painters)
```

```r
summary(School)
```

```r
summary(Composition)
```

`detach()` undoes it, after which the bare names no longer exist.

```r
detach(painters)
```

```r
tryCatch(
  summary(School),
  error = function(e) cat("Error:", conditionMessage(e), "\n")
)
```

Attaching is convenient but risky: if a variable of the same name already exists, it silently takes priority. `painters$School` is always unambiguous.

## subset()

```r
subset(painters, School == "F")
```

```r
painters[painters[["School"]] == "F", ]
```

## Combining data frames

Three techniques: `cbind()` places columns side by side, `merge()` joins on a common column, and `rbind()` stacks rows.

```r
df1
```

```r
df2
```

## cbind()

```r
cbind(df1, df2)
```

`state` appears twice. `cbind()` pastes columns together by position and knows nothing about what they mean.

---
## merge()

`merge()` joins two data frames on a common column, matching rows by value rather than by position.

```r
merge(df1, df2, by = "state")
```

`state` appears once, and the rows come back sorted alphabetically by the key.

### Why matching by value matters

Shuffle one data frame's rows. `cbind()` then pairs the wrong states together, while `merge()` still gets it right.

```r
df2_shuffled = df2[c(3, 1, 4, 2), ]
df2_shuffled
```

```r
cbind(df1, df2_shuffled)
```

```r
merge(df1, df2_shuffled, by = "state")
```

In the `cbind()` output, UP's population now sits beside AP's sample size. `merge()` is unaffected.

---
## rbind()

`rbind()` stacks two data frames vertically. The column names must match.

```r
df11 = data.frame(state = c("UP", "MP", "AP", "JK"), popnsize = c(1000, 2000, 3000, 4000))
df22 = data.frame(state = c("Bihar", "Delhi", "Punjab"), popnsize = c(100, 200, 300))
```

```r
df11
```

```r
df22
```

```r
rbind(df11, df22)
```

### Mismatched column names fail

```r
df33 = data.frame(region = c("Goa"), popnsize = c(50))
tryCatch(
  rbind(df11, df33),
  error = function(e) cat("Error:", conditionMessage(e), "\n")
)
```

## rbind() and cbind() on vectors

On plain vectors, mixing numbers and strings coerces everything to character.

```r
x = c("Apple", "Oranges", "Banana", "Apricot")
y = c(20, 30, 40, 50)
```

```r
rbind(x, y)
```

```r
cbind(y, x)
```

The numbers become `"20"`, `"30"` and so on. A data frame would have kept them numeric.

```r
data.frame(y, x)
```

## Summary

| Task | Command |
|---|---|
| Replace first match | `sub(old, new, x)` |
| Replace every match | `gsub(old, new, x)` |
| Indices of matches | `grep(pattern, x)` |
| Matching elements | `grep(pattern, x, value = TRUE)` |
| Ignore case | `grep(pattern, x, ignore.case = TRUE)` |
| TRUE/FALSE per element | `grepl(pattern, x)` |
| Create a data frame | `data.frame(a = , b = )` |
| Structure | `str(df)` |
| Row and column names | `rownames(df)`, `colnames(df)` |
| Extract a column | `df$col` |
| Refer to columns by name | `attach(df)` ... `detach(df)` |
| Filter rows | `subset(df, condition)` |
| Filter and drop columns | `subset(df, condition, select = c(-2, -4))` |
| Split into groups | `split(df, df$factor)` |
| Side by side | `cbind(df1, df2)` |
| Join on a key | `merge(df1, df2, by = "key")` |
| Stack | `rbind(df1, df2)` |

Four things worth carrying forward:

1. `sub()` replaces the first match, `gsub()` every match, and both match characters
   anywhere, including inside words.
2. `grep()` returns positions by default; add `value = TRUE` for the strings. `grepl()`
   returns TRUE/FALSE and has no `value` argument.
3. `cbind()` pairs rows by position; `merge()` pairs them by the key's value. Only
   `merge()` is safe when row order differs.
4. In `subset()` use `==` and quote factor levels: `School == "A"`, never `School = A`.
