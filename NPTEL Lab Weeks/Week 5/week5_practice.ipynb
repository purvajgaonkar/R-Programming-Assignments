## Missing data: NA

NA marks a value that is missing. Its position still exists in the vector.

```r
x <- c(10, 20, NA, 40, NA)
x
```

```r
length(x)
```

```r
is.na(x)
```

```r
sum(is.na(x))
```

```r
which(is.na(x))
```

### NA spreads through calculations

```r
sum(x)
```

```r
mean(x)
```

```r
sum(x, na.rm = TRUE)
```

```r
mean(x, na.rm = TRUE)
```

### Testing for NA the wrong way

Comparison with NA gives NA, not TRUE or FALSE. Always use `is.na()`.

```r
x == NA
```

```r
is.na(x)
```

### Removing or replacing missing values

```r
x[!is.na(x)]
```

```r
na.omit(x)
```

```r
y <- x
y[is.na(y)] <- 0
y
```

```r
complete.cases(x)
```

```r
x <- 10
if (x > 5) {
  print("x is greater than 5")
}
```

```r
x <- 3
if (x > 5) {
  print("this will not print")
}
print("done")
```

## if and else

```r
x <- 3
if (x > 5) {
  print("greater")
} else {
  print("not greater")
}
```

The `else` must sit on the same line as the closing brace, otherwise R treats the `if` as finished.

```r
marks <- 72
if (marks >= 50) {
  print("Pass")
} else {
  print("Fail")
}
```

---
## Nested if else if

```r
marks <- 72

if (marks >= 90) {
  grade <- "A"
} else if (marks >= 75) {
  grade <- "B"
} else if (marks >= 60) {
  grade <- "C"
} else if (marks >= 50) {
  grade <- "D"
} else {
  grade <- "F"
}

grade
```

```r
n <- 0

if (n > 0) {
  print("positive")
} else if (n < 0) {
  print("negative")
} else {
  print("zero")
}
```

### if works on a single value only

```r
v <- c(1, 5, 10)
tryCatch(
  if (v > 3) print("yes"),
  error = function(e) cat("Error:", conditionMessage(e), "\n")
)
```

With a vector, `if` uses only the first element in older R and raises an error in current R. Use `ifelse()` for vectors.

## ifelse for vectors

`ifelse(test, yes, no)` returns a vector the same length as the test.

```r
v <- c(1, 5, 10, 15)
ifelse(v > 5, "big", "small")
```

```r
ifelse(v %% 2 == 0, "even", "odd")
```

```r
scores <- c(45, 88, 67, 92, 30)
ifelse(scores >= 50, "Pass", "Fail")
```

### Nesting ifelse

```r
ifelse(scores >= 75, "High",
       ifelse(scores >= 50, "Medium", "Low"))
```

### ifelse and NA

```r
w <- c(1, NA, 5)
ifelse(w > 2, "yes", "no")
```

## switch

`switch()` picks one of several branches based on a value.

```r
day <- "tue"

switch(day,
  mon = "Monday",
  tue = "Tuesday",
  wed = "Wednesday",
  "Unknown day")
```

```r
switch("xyz",
  mon = "Monday",
  tue = "Tuesday",
  "Unknown day")
```

### switch on a number picks by position

```r
switch(2, "first", "second", "third")
```

### A small calculator

```r
calculate <- function(op, a, b) {
  switch(op,
    add = a + b,
    sub = a - b,
    mul = a * b,
    div = a / b,
    "unknown operation")
}

calculate("add", 10, 3)
calculate("mul", 10, 3)
calculate("pow", 10, 3)
```

## which

`which()` returns the positions where a condition is TRUE.

```r
v <- c(4, 9, 2, 15, 7, 20)
which(v > 8)
```

```r
v[which(v > 8)]
```

```r
which.max(v)
```

```r
which.min(v)
```

```r
which(v == 15)
```

### which with names

```r
named <- c(a = 4, b = 9, c = 2, d = 15)
which(named > 5)
```

### which on a matrix

```r
m <- matrix(1:9, nrow = 3)
which(m > 5)
```

```r
which(m > 5, arr.ind = TRUE)
```

### which and NA

```r
q <- c(3, NA, 8, 1)
which(q > 2)
```

`which()` ignores NA rather than returning it, which is why it is safer than raw logical indexing when missing values are present.

```r
q[q > 2]
```

```r
q[which(q > 2)]
```
